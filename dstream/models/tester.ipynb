```python
from dstream.dataset import classification_dataset

df = classification_dataset()
features = df.columns[:-1]
target = 'target'

from dstream.preprocess import DataTransformer

preprocessor = DataTransformer.from_default()

X_train, X_test, y_train, y_test = preprocessor.run(data=df, features=features, target=target)
```

```python
from dstream.models.trainer import MLTrainer
```

```python
trainer = MLTrainer.from_classification(train_dataset=(X_train, y_train))
```

```python
trainer.train()
```

```python
y_pred = trainer.predict(X_test)
```

```python
trainer.evaluate(eval_dataset=(X_test, y_test))
```

```python
trainer.save()
```
